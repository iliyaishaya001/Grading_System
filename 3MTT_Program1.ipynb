# Program Title  
User Greeting and Age Projection Program

### Description  
This program prompts the user to enter their name and age, then displays a personalized greeting along with their projected age in 10 years.

### Objective  
To demonstrate:
User input handling
Type conversion (string → integer)
Function definition and usage
Basic string formatting in Python

### How the Program Works  
The program asks the user to input their name.  
It then requests the user's current age.  
The age input is converted from a string to an integer.  
The program calculates the user’s age after 10 years.  
A formatted message is displayed to the user.  

```python
def greet():
    name = input("Please type your name: ")
    age = int(input("Please enter your age: "))
    print(f'Hello {name}! In 10 years years you will be {age + 10} years old.')


greet()
```
